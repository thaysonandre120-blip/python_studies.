'''Enunciado

Crie um programa em Python que simule o funcionamento de um caixa eletrônico simples.

O programa deve:

Iniciar com um saldo de R$ 500,00.
Exibir um menu de opções para o usuário, repetidamente, até que ele escolha sair:
   1 - Consultar saldo
   2 - Depositar
   3 - Sacar
   4 - Sair
Ler a opção escolhida pelo usuário via teclado (input()).
Executar a ação correspondente:
Consultar saldo: exibir o saldo atual formatado (ex: Saldo atual: R$ 500.00).
Depositar: solicitar um valor e somá-lo ao saldo.
Sacar: solicitar um valor;
se for maior que o saldo disponível, exibir a mensagem "Saldo insuficiente." e não realizar o saque;
caso contrário, subtrair o valor do saldo.
Sair: exibir a mensagem "Até logo!" e encerrar o programa.
O programa deve continuar exibindo o menu após cada operação, até que o usuário escolha a opção Sair.
Requisitos técnicos
Utilizar uma estrutura de repetição (while) para manter o menu ativo.
Utilizar estruturas condicionais (if / elif / else) para tratar as opções do menu.
Utilizar input() para capturar as escolhas e valores do usuário.
Utilizar print() formatado para exibir saldo e mensagens.
Desafios extras (opcional)
Impedir depósitos ou saques com valores negativos ou iguais a zero.
Cobrar uma taxa de R$ 2,00 em toda operação de saque.
Limitar a 3 tentativas consecutivas de opção inválida no menu; após isso, encerrar o programa automaticamente.
Exemplo de execução esperada
=== CAIXA ELETRÔNICO ===
1 - Consultar saldo
2 - Depositar
3 - Sacar
4 - Sair
Escolha uma opção: 1
Saldo atual: R$ 500.00

=== CAIXA ELETRÔNICO ===
1 - Consultar saldo
2 - Depositar
3 - Sacar
4 - Sair
Escolha uma opção: 3
Digite o valor do saque: 700
Saldo insuficiente.#
